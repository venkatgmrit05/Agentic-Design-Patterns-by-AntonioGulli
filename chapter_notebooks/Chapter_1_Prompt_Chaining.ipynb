```python
#about langchain ()

# LangChain Expression Language (LCEL) is a declarative syntax designed to simplify the composition of chains within the LangChain framework. It allows developers to build complex AI workflows by piping components together using a minimalist code layer. 
# LangChain Blog
# LangChain Blog
#  +3
# Core Syntax and Concepts
# The Pipe Operator (|): Similar to a Unix pipe, it connects LangChain components (Prompts, Models, Parsers) so that the output of one step becomes the input of the next.
# Runnable Interface: The fundamental building block of LCEL. Every component in a chain is a "Runnable," which implements a standard interface for various execution modes.
# Declarative Composition: Instead of imperative code, you describe what the chain should do (e.g., chain = prompt | model | parser), which LangChain then optimizes for execution. 
# LangChain Blog
# LangChain Blog
#  +5
# Key Features and Benefits
# Chains built with LCEL automatically support advanced production features without additional code changes: 
# LangChain Blog
# LangChain Blog
#  +1
# Unified Execution Modes: Every chain supports invoke() (single call), batch() (parallel inputs), and stream() (token-by-token output).
# Asynchronous Support: All chains expose async methods (ainvoke, astream) for high-throughput applications.
# Parallel Execution: Components can run concurrently using RunnableParallel, reducing latency for tasks like multiple retrieval steps.
# Observability: Integrated with LangSmith to provide automatic tracing, debugging, and visualization of every step in the chain.
# Reliability: Built-in support for retries and fallbacks, allowing the system to switch models if a primary call fails. 
# LangChain Blog
# LangChain Blog
#  +8
# Common LCEL Primitives
# Component 	Description
# RunnablePassthrough	Forwards input unchanged or adds extra keys to a dictionary.
# RunnableParallel	Executes multiple runnables in parallel and returns their outputs as a dictionary.
# RunnableLambda	Wraps a standard Python function to make it compatible with LCEL piping.
# RunnableBranch	Implements conditional logic to route data to different chains based on input.
# Comparison: Traditional vs. LCEL
# While traditional LangChain used specialized classes like LLMChain, LCEL is the preferred modern method because it is more transparent. 
# LangChain Blog
# LangChain Blog
#  +1
# Traditional: Prompts were often "hidden" inside library code, making them harder to customize for production.
# LCEL: Prompts are prominent and easily swappable, making the entire pipeline modular and easier to debug. 
# https://medium.com/@mustafa_akca/a-beginners-guide-to-langchain-expression-language-lcel-building-powerful-chains-b1711529efd2
```

```python
import os
# from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_community.llms import Ollama

# For better security, load environment variables from a .env file
# from dotenv import load_dotenv
# load_dotenv()
# Make sure your OPENAI_API_KEY is set in the .env file

# Initialize the Language Model (using ChatOpenAI is recommended)
# llm = ChatOpenAI(temperature=0)
# llm = Ollama(model="llama3")
llm = Ollama(model="gemma4")

# --- Prompt 1: Extract Information ---
prompt_extract = ChatPromptTemplate.from_template(
    "Extract the technical specifications from the following text:\n\n{text_input}"
)

# --- Prompt 2: Transform to JSON ---
prompt_transform = ChatPromptTemplate.from_template(
    "Transform the following specifications into a JSON object with 'cpu', 'memory', and 'storage' as keys:\n\n{specifications}"
)

# --- Build the Chain using LCEL ---
# chain = prompt | model | parser

# The StrOutputParser() converts the LLM's message output to a simple string.
extraction_chain = prompt_extract | llm | StrOutputParser()

# The full chain passes the output of the extraction chain into the 'specifications'
# variable for the transformation prompt.
full_chain = (
    {"specifications": extraction_chain}| prompt_transform| llm | StrOutputParser()
)
```

```python
print(prompt_extract,'\n')
print(prompt_transform)
```

```python

# --- Run the Chain ---
input_text = "The new laptop model features a 3.5 GHz octa-core processor, 16GB of RAM, and a 1TB NVMe SSD."

# Execute the chain with the input text dictionary.
final_result = full_chain.invoke({"text_input": input_text})

print("\n--- Final JSON Output ---")
print(final_result)
```
